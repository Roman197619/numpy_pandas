* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
# your code
x = [2, 2, 2, 3, 3, 3, 5]

nums = []
counts = []

curr_num = x[0]
curr_count = 0

for i in range(len(x)):
    curr_count += 1
    if i + 1 == len(x) or x[i] != x[i+1]:
        nums.append(x[i])
        counts.append(curr_count)
        curr_count = 0

print(nums, counts)   
```

```python
x = np.array([2, 2, 2, 3, 3, 3, 5])


change_indices = np.where(np.diff(x) != 0)[0]

nums = x[np.append(0, change_indices + 1)]

full_indices = np.concatenate([[-1], change_indices, [len(x) - 1]])

counts = np.diff(full_indices)

print(nums, counts)
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
import numpy as np
from scipy.spatial.distance import cdist
import time

rng = np.random.default_rng()
x = rng.random(size=(500, 20))
y = rng.random(size=(700, 20))

def pairwise_dist_py(X_list, Y_list):
    out = []
    for x_row in X_list:
        row = []
        for y_row in Y_list:
            s = 0.0
            for xi, yi in zip(x_row, y_row):
                d = xi - yi
                s += d * d
            row.append(s ** 0.5)
        out.append(row)
    return out

def pairwise_dist_np(X_arr, Y_arr):
    return np.sqrt(((X_arr[:, None, :] - Y_arr[None, :, :]) ** 2).sum(axis=2))

def pairwise_dist_fast_np(X, Y):
    x2 = np.sum(X**2, axis=1)[:, np.newaxis]
    y2 = np.sum(Y**2, axis=1)[np.newaxis, :]
    xy = X @ Y.T
    return np.sqrt(np.maximum(x2 - 2*xy + y2, 0))

t0 = time.perf_counter()
res_py = pairwise_dist_py(x.tolist(), y.tolist())
t_py = time.perf_counter() - t0

t0 = time.perf_counter()
res_np = pairwise_dist_np(x, y)
t_np = time.perf_counter() - t0

t0 = time.perf_counter()
res_fast_np = pairwise_dist_fast_np(x, y)
t_fast_np = time.perf_counter() - t0

t0 = time.perf_counter()
res_cdist = cdist(x, y, metric="euclidean")
t_cdist = time.perf_counter() - t0

print(f"Equal results: {np.allclose(res_py, res_np, res_fast_np, res_cdist)}")

print(f"Python loops:      {t_py:.4f} s")
print(f"Numpy vectorized:  {t_np:.4f} s")
print(f"Numpy Matrix Mult: {t_fast_np:.4f} s")
print(f"Scipy cdist:       {t_cdist:.4f} s")
```
