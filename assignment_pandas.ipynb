**4-5. Рассчитайте среднее значение и среднеквадратичное отклонение возраста тех, кто получает более 50K в год (признак *salary*) и тех, кто получает менее 50K в год**

**6. Правда ли, что люди, которые получают больше 50k, имеют минимум высшее образование? (признак *education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters* или *Doctorate*)**

```python
higher = ['Bachelors','Prof-school','Masters','Doctorate']

rich_education = data.loc[data['salary'] == '>50K', 'education']

is_higher = rich_education.isin(higher)

all_higher = is_higher.all()      
share_higher = is_higher.mean() 

print(all_higher)
print(f"доля высшего образования среди богатых: {share_higher:.2f}")
```

**7. Выведите статистику возраста для каждой расы (признак *race*) и каждого пола. Используйте *groupby* и *describe*. Найдите таким образом максимальный возраст мужчин расы *Asian-Pac-Islander*.**

```python
result = data.groupby(['race', 'sex']).describe()
print(result)
print(f"Максимальный возраст среди Asian-Pac-Islander мужского пола: {result.loc[('Asian-Pac-Islander', 'Male'), ('age', 'max')]}")
```

**8. Среди кого больше доля зарабатывающих много (>50K): среди женатых или холостых мужчин (признак *marital-status*)? Женатыми считаем тех, у кого *marital-status* начинается с *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), остальных считаем холостыми.**

```python
married = ['Married-civ-spouse', 'Married-AF-spouse', 'Married-spouse-absent']

is_married = data['marital-status'].isin(married)

res = (data['salary'] == '>50K').groupby(is_married).mean()

res_idx = res.idxmax()
res_label = 'married' if res_idx else 'unmarried'

print(res_label, res[res_idx])
```

**9. Какое максимальное число часов человек работает в неделю (признак *hours-per-week*)? Сколько людей работают такое количество часов и каков среди них процент зарабатывающих много?**

```python
max_work_hours = data['hours-per-week'].max()

is_max_worker = data['hours-per-week'] == max_work_hours

max_workers_salaries = data.loc[is_max_worker, 'salary']

max_workers_count = len(max_workers_salaries)
rich_percentage = (max_workers_salaries == '>50K').mean() * 100

print(f"Максимальное количество рабочих часов: {max_work_hours}")
print(f"Количество работников, работающих максимальное количество часов: {max_workers_count}")
print(f"Процент зарабатывающих >50K среди них: {rich_percentage:.2f}%")
```

**11.Сгруппируйте людей по возрастным группам *young*, *adult*, *retiree*, где:**
* *young* соответствует 16-35 лет
* *adult* - 35-70 лет
* *retiree* - 70-100 лет

**Проставьте название соответсвтуещей группы для каждого человека в новой колонке AgeGroup**

```python
data['AgeGroup'] = pd.cut(data['age'], bins=[15, 35, 70, 100], labels=['young', 'adult', 'retiree'])

data.head(5)
```

**12-13. Определите количество зарабатывающих >50K в каждой из возрастных групп (колонка AgeGroup), а также выведите название возрастной группы, в которой чаще зарабатывают больше 50К (>50K)**

```python
age_analysis = (
    (data['salary'] == '>50K')
    .groupby(data['AgeGroup'], observed=True)
    .agg(['sum', 'mean']) 
    .rename(columns={'sum': 'count >50K', 'mean': 'share >50K'})
)

print(age_analysis)
best_group = age_analysis['share >50K'].idxmax()
best_percentage = age_analysis['share >50K'].max()

print(f"\nГруппа, в которой чаще всего зарабатывают >50K: {best_group}")
print(f"Доля >50K в этой группе: {best_percentage:.2%}")
```

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функциюю фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
grouped = (
    data
    .groupby('occupation')
    .agg(mean_age=('age','mean'), min_hours=('hours-per-week','min'), count=('age','size'))
)

print(grouped.loc[:, ['count']])

def filter_func(x):
    return (x['mean_age'] <= 40) & (x['min_hours'] > 5).all()

filtered = grouped[grouped.apply(filter_func, axis=1)]

print(filtered)
```
